```python
import numpy as np

A = [[1, 1, 1],
     [3, 2, 1],
     [2, 1, 2]]

Ainv = np.linalg.inv(A)
print("Inverce matrix A:")
print(Ainv)
```

```python
import numpy as np

A = [[4, 6, 2],
     [3, 4, 1],
     [2, 8, 13]]

s = [9, 7, 2]

r = np.linalg.solve(A, s)
print("A*r=s solution. \nr equals to")
print(r)
```
